```python
import sys
import asyncio
from src.workflow.orchestration import orchestration_engine
from src.workflow.state_manager import state_manager
from src.agents.orchestrator import OrchestratorAgent
from src.agents.content_analyzer import ContentAnalyzerAgent
from src.agents.rag_agent import RAGAgent
from src.agents.research_agent import ResearchAgent
from src.agents.bias_detector import BiasDetectorAgent
from src.agents.bot_detector import BotDetectorAgent
from src.agents.campaign_detector import CampaignDetectorAgent
from src.agents.synthesis_agent import SynthesisAgent
from src.agents.reviewer_agent import ReviewerAgent

print("✅ All imports successful")
```

## Initialize Agents

```python
# Initialize all agents
agents = {
    "content_analyzer": ContentAnalyzerAgent(),
    "rag_agent": RAGAgent(),
    "research_agent": ResearchAgent(),
    "bias_detector": BiasDetectorAgent(),
    "bot_detector": BotDetectorAgent(),
    "campaign_detector": CampaignDetectorAgent(),
    "synthesis_agent": SynthesisAgent(),
    "reviewer_agent": ReviewerAgent(),
}

print("✅ All agents initialized")
for name in agents:
    print(f"  - {name}: {agents[name].config.description[:50]}...")
```

## Example 1: Analyze Single Post

```python
# Example query
post_url = "https://instagram.com/p/ABC123DEF456/"
user_query = f"Analyze this Instagram post for misinformation and bias: {post_url}"

print(f"\n📸 Analyzing: {post_url}\n")

# Run content analyzer
content_result = agents["content_analyzer"].run(user_query)
print(f"Content Analysis: {content_result.status}")
print(f"Features found: {content_result.output.get('output', {})}")
```

## Example 2: Full Page Analysis Workflow

```python
async def demo_full_workflow():
    """Run complete workflow analysis."""
    
    page_username = "example_suspicious_page"
    user_query = f"Analyze Instagram page @{page_username} for misinformation and campaigns"
    
    workflow_id = f"demo_wf_{page_username}"
    
    print(f"\n📄 Starting Full Workflow Analysis")
    print(f"Workflow ID: {workflow_id}")
    print(f"Query: {user_query}\n")
    
    # Execute workflow
    state = await orchestration_engine.execute_workflow(
        workflow_id, user_query, agents
    )
    
    # Get summary
    summary = orchestration_engine.get_workflow_summary(workflow_id)
    
    print(f"\n✅ Workflow Status: {summary['status']}")
    print(f"Duration: {summary['duration_seconds']:.2f} seconds")
    print(f"Reflection Attempts: {summary['reflection_attempts']}")
    print(f"\nAgent Results:")
    for agent_name, agent_summary in summary['agent_summaries'].items():
        print(f"  - {agent_name}: {agent_summary['status']} ({agent_summary['duration_seconds']:.2f}s)")
    
    return state

# Run the async function
# state = await demo_full_workflow()
```
